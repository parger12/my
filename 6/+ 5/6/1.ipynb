```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

data = pd.read_csv(r'E:\courses\Data science\my\6\5\flights_preprocessed.csv')

target = data['Arrival Delay']
features = data.drop(['Arrival Delay'] , axis=1)
features_train, features_valid, target_train, target_valid = train_test_split(
    features, target, test_size=0.25, random_state=12345) 

model = RandomForestRegressor(random_state=12345, n_estimators=60, max_depth=11)
model.fit(features_train, target_train) 
print("Наилучшая модель")
print("R2 на обучающей выборке: ", model.score(features_train, target_train)) # найдите значение метрики R2 на обучающей выборке
print("R2 на валидационной выборке: ", model.score(features_valid, target_valid)) # найдите значение метрики R2 на валидационной выборке
```
